```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering, Divise
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
```

```python
data = {
    "x" : [2.5, 0.5, 2.2, 1.9, 3.1],
    "y" : [2.4, 0.7, 2.9, 2.2, 3.0]
}

call = pd.DataFrame(data)
```

```python
call
```

```python
linker = linkage(call)
linker
```

```python
plt.figure(figsize=(8, 4))
dendrogram(linker)
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Distance')
plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=3) # after watching dendogram then decide how many clusters you want to make 
```

```python
model.fit_predict(call)
```

```python
call["Clusters"] = model.fit_predict(call)
call
```
