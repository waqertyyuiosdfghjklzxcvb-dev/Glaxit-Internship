```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'age' : [12,34,23,34,56,45,56,34],
    'spend' : [2000,30000,40000,5000,5000,20000,10000,12000]
}

frame = pd.DataFrame(data)
```

```python
frame.head(10)
```

```python
scaler = StandardScaler()
```

```python
scaling = scaler.fit_transform(frame)
```

```python
model = KMeans(n_clusters = 2,random_state = 42)
```

```python
m = model.fit(scaling)
print(m)
```

```python
labels = m.labels_

i = len(labels)

add= []
for j in range(0,i):
    add.append("Cluster 1" if labels[j] == 0 else "Cluster 2")
```

```python
# Metrics
```

```python
print(silhouette_score(frame,labels))
```

```python
print(davies_bouldin_score(frame,labels))
```

```python
print(calinski_harabasz_score(frame,labels))
```

```python
frame["Cluster"] = add
frame
```
