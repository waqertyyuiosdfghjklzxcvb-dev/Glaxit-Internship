```python
import pandas as pd
from kmedoids import KMedoids
import gower
```

```python
data = {
    'name':       ['Ali', 'Sara', 'John', 'Fatima', 'Mike', 'Zara', 'Tom', 'Hina', 'Jake', 'Nadia'],
    'gender':     ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'blood_group':['A+', 'B+', 'O+', 'A+', 'AB+', 'O-', 'B+', 'A-', 'O+', 'AB+'],
    'city':       ['Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Islamabad', 'Karachi'],
    'education':  ['Matric', 'Bachelor', 'Master', 'Bachelor', 'PhD', 'Matric', 'Master', 'Bachelor', 'PhD', 'Matric']
}
```

```python
frame = pd.DataFrame(data)
```

```python
cat_cols = ['gender', 'blood_group', 'city', 'education']
frame[cat_cols] = frame[cat_cols].astype(object)
```

```python
D = gower.gower_matrix(frame[['gender', 'blood_group', 'city', 'education']])
print(D)
```

```python
model = KMedoids(n_clusters=2,random_state= 42)
```

```python
fit = model.fit(D)
```

```python
length = len(fit.labels_)
length
label = fit.labels_
```

```python
lis = []
for i in range(10):
    lis.append("Cluster 1" if label[i] == 0 else "Cluster 2")
```

```python
frame["Cluster"] = lis
```
