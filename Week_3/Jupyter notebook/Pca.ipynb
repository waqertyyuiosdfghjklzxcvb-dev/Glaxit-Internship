```python
from sklearn.decomposition import PCA
import pandas as pd
```

```python
data = {
    "x" : [2.5,0.5,2.2,1.9,3.1],
    "y" : [2.4,0.7,2.9,2.2,3.0]
}
```

```python
call = pd.DataFrame(data)
```

```python
call
```

```python
model = PCA(n_components=1,random_state=42)
```

```python
result = model.fit_transform(call)
```

```python
print(model)
```

```python
print(f"Variance Ratio : ",model.explained_variance_ratio_) # how much variance is explained
```

```python
print(model.components_) # eigenvector (principal component direction)
```

```python
model.mean_ 
```

```python
call["PC"] = result
```
